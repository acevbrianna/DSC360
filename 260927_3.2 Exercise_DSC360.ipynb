Importing "pytesseract" helps so Python can communicate with the Tesseract OCR program. I am also going to import image from the Pillow library so that Python can open the Bowers image. I will also import spaCy so I can perform natural language processing on the text that is extracted from the image.

```python
import pytesseract
from PIL import Image
import spacy
```

```python
pytesseract.pytesseract.tesseract_cmd = (
    r"C:\Program Files\Tesseract-OCR\tesseract.exe"
)
```

```python
#test
print(pytesseract.get_tesseract_version())
```

I tested the location of Tesseract after it was installed. It was successful.

```python
#OCR Code
from PIL import Image

image = Image.open("bowers.jpg")

bowers_text = pytesseract.image_to_string(image)

print(bowers_text)
```

```python
#use SpaCy
!python -m spacy download en_core_web_sm
```

Had to successfully install the package for SpaCy to work. Now we should be good to go. 

```python
#Wanted to use the english language model for SpaCy
nlp = spacy.load("en_core_web_sm")

doc = nlp(bowers_text)

for token in doc:
    print(token.text, token.pos_, token.dep_)
```

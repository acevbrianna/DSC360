First, I will make a request to the base API and view the response as a Python dictionary. I will then get the activity's API key from the response and use that key to request the specific activity again. The Python API Tutorial from Dataquest explains that APIs allow programs to request information from a server. Python's requests library can be used to make a GET request, and the .json() method can convert the returned JSON data into a Python object that is easier to work with.

```python
#import library
import requests
```

I will start by using the base Bored API without adding any extra information. Since I am not giving the API any parameters, it should return something random. I will also check the status code to make sure the request was successful.

```python
#make the API request
#bored API base URL
url = "https://bored.api.lewagon.com/api/activity"

#make a GET request
response = requests.get(url)

#check the status code
print("Status Code:", response.status_code)
```

The status code will tell me whether my request worked. A status code of 200 means the API successfully received my request and returned a response.

```python
#convert to a dictionary
activity_data = response.json()

activity_data
```

The API returned the activity information as JSON, which Python can represent as a dictionary when converted. The dictionary contains several key-value pairs. And since I converted it as a Python dictionary, I can access individual values by using their keys.

```python
#get the key from the dictionary
activity_key = activity_data["key"]

print("Activity Key:", activity_key)
```

I pulled the "key" value from the dictionary and stored it in a new variable called "activity_key". I can now send that key back to the API as a parameter. Instead of asking for another random activity, this request should return the activity connected to that specific key.

```python
#use the key to request a specific activity
specific_response = requests.get(
    url,
    params={"key": activity_key}
)

print("Status Code:", specific_response.status_code)

specific_activity = specific_response.json()
specific_activity
```

For the specific activity request, I used the activity key as a query parameter. This allows me to request the specific activity connected to the key instead of receiving another random activity. Next, I will access each value in the returned dictionary separately so the final output is easier to read.

```python
#display each attribute
print("Activity:", specific_activity["activity"])
print("Key:", specific_activity["key"])
print("Link:", specific_activity["link"])
print("Participants:", specific_activity["participants"])
print(f"Price: ${specific_activity['price']:.2f}")
print("Type:", specific_activity["type"].title())
print("Accessibility:", specific_activity["accessibility"])
```

This exercise helped me understand how Python can communicate with an API and work with the information it returns. The first request used the base API to retrieve a random activity. After converting the JSON response into a Python dictionary, I was able to access the "key" value directly. I then used that key as a parameter in a second API request to retrieve the specific activity. Finally, I was able to access each value in the returned dictionary and formatted the information so it was easier to read. One thing that stood out to me is how similar JSON data is to a Python dictionary. Once the response was converted with .json(), I could access information such as the activity, price, participants, and key by using their names. This made the API response much easier to work with. It's super interesting to see how the two can interact with one another efficiently.
